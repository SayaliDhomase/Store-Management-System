# Store Management System Documentation

## Overview:

This is a simple Store Management System application built using Python's Tkinter library. It allows users to manage a store's inventory, including adding items, deleting items, viewing the current stock, and selling items.

## Features:

1. **Show Items:**
   - Displays the current items in the store along with their quantities.
   - Click the "VIEW STOCK" button to see the current inventory.

    ```python
    def show_items():
        m = ''
        for i in stock:
            m += f"{i['item']} - quantity {i['quantity']}\n"
        messagebox.showinfo(title='items', message=m)
    ```

2. **Add Items:**
   - Adds new items to the store's inventory.
   - Enter the item name, quantity, and price, then click the "ADD ITEM" button.

    ```python
    def add_items():
        items = Entry_1.get()
        flag = 0
        q = Entry_2.get()
        p = Entry_3.get()
        for i in range(len(stock)):
            if items == stock[i]['item']:
                stock[i]['quantity'] += q
                flag = 1
                break
        if flag == 0:
            itemss = {'item': items, 'quantity': q, 'price': p}
            stock.append(itemss)
        Entry_1.delete(0, len(Entry_1.get()))
        Entry_2.delete(0, len(Entry_2.get()))
        Entry_3.delete(0, len(Entry_3.get()))
        messagebox.showinfo(title='items added', message=f"{items} added")
    ```

3. **Sell Items:**
   - Decreases the quantity of a selected item when sold.
   - Enter the item name and the quantity to be sold, then click the "SELL ITEM" button.

    ```python
    def sell_items(item_to_sell, Quantity):
        Quantity = int(Quantity)
        items_sold = {}
        items_sold[item_to_sell] = Quantity
        for i in stock:
            if item_to_sell in i.values():
                i['quantity'] -= Quantity
        for i, j in items_sold.items():
            messagebox.showinfo(title='Bill', message=f"{i} sold, qty: {j}")
        Entry_1.delete(0)
        Entry_2.delete(0)
        Entry_3.delete(0)
    ```

4. **Delete Item:**
   - Removes an item from the inventory.
   - Enter the name of the item to delete and click the "DELETE ITEM" button.

    ```python
    def delete_item():
        item_to_delete = Entry_1.get()
        for i in stock:
            for key, value in i.items():
                if item_to_delete == value:
                    stock.remove(i)
        Entry_1.delete(0, len(Entry_1.get()))
        Entry_2.delete(0, len(Entry_2.get()))
        Entry_3.delete(0, len(Entry_3.get()))
        messagebox.showinfo(title='deleted', message=f'{item_to_delete} deleted successfully')
    ```

5. **Clear Screen:**
   - Clears the input fields on the screen.
   - Click the "CLEAR SCREEN" button to reset the input fields.

    ```python
    def Clear_Item():
        Entry_1.delete(0, len(Entry_1.get()))
        Entry_2.delete(0, len(Entry_2.get()))
        Entry_3.delete(0, len(Entry_3.get()))
    ```

6. **Exit:**
   - Closes the application.
   - Click the "Exit" button to close the program.

    ```python
    def Exit():
        root.destroy()
    ```
